### 1. Contexto Estratégico y Problema de Negocio
La seguridad alimentaria y la estabilidad de precios dependen críticamente de los flujos de carga origen-destino. Este dataset permite caracterizar la matriz de origen municipal y recepción urbana bajo la teoría de redes de distribución agroalimentaria.

**Objetivo de Negocio**:
> Monitorear la dinámica de abastecimiento mayorista de alimentos en Colombia para detectar riesgos de desabastecimiento, cuellos de botella logísticos y asimetrías de oferta territorial.

### 2. Entidad del Dominio y Alcance
* **Entidad Analítica Principal**: `Flujos de Carga Agroalimentaria (Kg ingresados por Origen-Destino)`
* **Custodio Oficial de los Datos**: Departamento Administrativo Nacional de Estadística (DANE) - Dirección de Metodología y Producción Estadística (DIMPE)
* **Granularidad Espacial**: `Bidireccional: Municipio Origen (DIVIPOLA) -> Central Mayorista Destino`
* **Granularidad Temporal**: `Diaria (Días hábiles de mercado)`

### 3. Preguntas de Negocio Mapeadas (`docs/1-Bateria_preguntas.md`):
* **D1: Brecha Oferta-Mercado**
* **D2: Persistencia del Gap Estacional**
* **I2: Relación Abastecimiento-Precio**

### 4. Indicadores Clave de Desempeño (KPIs):
* `Volumen total abastecido (Toneladas)`
* `Concentración de orígenes (HHI)`
* `Índice de brecha oferta-mercado`

### 5. Fundamentación del Contrato de Datos (Data Contract)

1. El custodio del dato y la periodicidad de actualización.
2. La definición operativa de las variables críticas y sus tipos de datos esperados.
3. Las tolerancias de calidad (porcentaje máximo de nulos y umbrales de detección de valores anómalos).
