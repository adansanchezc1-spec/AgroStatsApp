### 1. Contexto Estratégico y Problema de Negocio
La teoría de transmisión de precios analiza la asimetría en el traspaso de costos a lo largo de la cadena. Un descalce prolongado entre IPP e IPC evidencia concentración o ineficiencias en la intermediación comercial.

**Objetivo de Negocio**:
> Evaluar la transmisión vertical de precios desde el productor agropecuario (IPP) hasta la canasta básica familiar urbana (IPC Alimentos).

### 2. Entidad del Dominio y Alcance
* **Entidad Analítica Principal**: `Índice de Precios al Consumidor (IPC Alimentos) e Índice de Precios del Productor (IPP Agropecuario)`
* **Custodio Oficial de los Datos**: DANE - Dirección de Metodología y Producción Estadística (DIMPE) / Subdirección de Precios
* **Granularidad Espacial**: `Nacional / 23 Ciudades Capitales`
* **Granularidad Temporal**: `Mensual`

### 3. Preguntas de Negocio Mapeadas (`docs/1-Bateria_preguntas.md`):
* **A1: Variación del Índice Macroeconómico**
* **G1: Margen Bruto Comercial**
* **G3: Transmisión Vertical de Precios**

### 4. Indicadores Clave de Desempeño (KPIs):
* `Inflación anualizada de alimentos`
* `Spread IPP vs. IPC`
* `Elasticidad de transmisión vertical`
